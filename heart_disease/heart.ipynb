```python
rbp_mean = df.loc[df['RestingBP'] != 0,'RestingBP'].mean()
df["RestingBP"]=df["RestingBP"].replace(0,rbp_mean)
df["RestingBP"]=df["RestingBP"].round(2)
```

```python
df_encoded=pd.get_dummies(df,drop_first=True)
df_encoded
```

```python
df_encoded = df_encoded.astype(int)
df_encoded.head()
```

```python
from sklearn.preprocessing import StandardScaler
numerical_cols = ['Age', 'RestingBP', 'Cholesterol', 'MaxHR', 'Oldpeak']
scaler = StandardScaler()
df_encoded[numerical_cols] = scaler.fit_transform(df_encoded[numerical_cols])
df_encoded.head()
```

```python
from sklearn.model_selection import train_test_split
x=df_encoded.drop("HeartDisease",axis=1)
y=df_encoded["HeartDisease"]
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(x_train,y_train)
```

```python
from sklearn.metrics import accuracy_score
pred = model.predict(x_test)
print(accuracy_score(y_test,pred))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
```

```python
model1=KNeighborsClassifier()
model2=SVC()
model3=DecisionTreeClassifier()
model4=LogisticRegression()
model5=GaussianNB() 
```

```python
model1.fit(x_train,y_train)
model2.fit(x_train,y_train)
model3.fit(x_train,y_train)
model4.fit(x_train,y_train)
model5.fit(x_train,y_train)
```

```python
y_pred1=model1.predict(x_test)
y_pred2=model2.predict(x_test)
y_pred3=model3.predict(x_test)
y_pred4=model4.predict(x_test)
y_pred5=model5.predict(x_test)
```

```python
print("KNN Accuracy:",accuracy_score(y_test,y_pred1))
print("SVC Accuracy:",accuracy_score(y_test,y_pred2))
print("Decision Tree Accuracy:",accuracy_score(y_test,y_pred3))
print("Logistic Regression Accuracy:",accuracy_score(y_test,y_pred4))
print("GaussianNB Accuracy:",accuracy_score(y_test,y_pred5))
```
