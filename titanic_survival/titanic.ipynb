```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB

from sklearn.preprocessing import StandardScaler

from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

```python
train=pd.read_csv("titanic_train.csv")
test=pd.read_csv("titanic_test.csv")
```

```python
train.isnull().sum()
```

```python
train=train.dropna(subset=["Embarked"])
```

```python
test.isnull().sum()
```

```python
train.drop("Cabin", axis=1, inplace=True)
test.drop("Cabin", axis=1, inplace=True)
```

```python
train.isnull().sum()
```

```python
test=test.fillna(test["Fare"].mean())
```

```python
test.drop(["Name","Ticket"], axis=1, inplace=True)
```

```python
plt.figure(figsize=(10,6))
sns.histplot(x="Age", hue="Survived",data=train,kde=True)
plt.show
```

```python
train = train.fillna(train["Age"].mean())
test = test.fillna(test["Age"].mean())
```

```python
train.head()
```

```python
train["Embarked"].value_counts()
```

```python
x = train.drop(["Survived","Name","Ticket",], axis=1)
y = train["Survived"]
```

```python
x

```

```python
x_train=pd.get_dummies(x,drop_first=True,dtype=int)
test=pd.get_dummies(test,drop_first=True,dtype=int)
x_train.head()
```

```python
scaler=StandardScaler()
x_train[["Age","Fare"]]=scaler.fit_transform(x_train[["Age","Fare"]])
test[["Age","Fare"]]=scaler.fit_transform(test[["Age","Fare"]])
x_train.head()
```

```python
lr=LogisticRegression(max_iter=1000)
lr.fit(x_train,y)
```

```python
test.head()
```

```python
y_pred=lr.predict(test)
```

```python
y_test=pd.read_csv("datasets/titanic_test_result.csv")
accuracy_score(y_test["Survived"],y_pred)
```

```python
knn=KNeighborsClassifier()
knn.fit(x_train,y)
```

```python
y_pred=knn.predict(test)
accuracy_score(y_test["Survived"],y_pred)
```

```python
nb=GaussianNB()
nb.fit(x_train,y)
```

```python
y_pred=nb.predict(test)
accuracy_score(y_test["Survived"],y_pred)
```

```python
svc=SVC()
svc.fit(x_train,y)
```

```python
y_pred=svc.predict(test)
accuracy_score(y_test["Survived"],y_pred)
```

```python
desicion=DecisionTreeClassifier()
desicion.fit(x_train,y)
```

```python
y_pred=desicion.predict(test)
accuracy_score(y_test["Survived"],y_pred)
```
